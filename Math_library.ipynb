```python
#1.Write a program to calculate the cross sectional area, surface area, volume of a cylinder.
import math as m
r=5   #Radius of the cylinder in cm
h=7   #Height of the cylinder in cm
carea=m.pi*r*r       #cross sectional area of the cylinder
sarea=2*m.pi*r*h     #surface area of the cylinder
volume=m.pi*r*r*h    #volume of the cylinder
print(f"Cross sectional area of the cylinder is {carea:.2f} sq. cm")
print(f"Surface area of the cylinder is {sarea:.2f} sq. cm")
print(f"Volume of the cylinder is {volume:.2f} cc")
```

```python
#16.Use math library to find horizontal and vertical components of an inclined Force.

```

```python
import math as m

# Define the force and angle
force = 100  # Force in Newtons
angle_degrees = 30  # Angle in degrees

# Convert angle to radians
angle_radians = m.radians(angle_degrees)

# Calculate horizontal and vertical components
horizontal_component = force * m.cos(angle_radians)
vertical_component = force * m.sin(angle_radians)

# Print the results
print(f"Force: {force} N")
print(f"Angle: {angle_degrees} degrees")
print(f"Horizontal component: {horizontal_component:.2f} N")
print(f"Vertical component: {vertical_component:.2f} N")
```
