```python
#Write a code to separate digits of a three digits number
N=157           #157 is a three digits number
a=N//100        #a is the digit at 100th place
print('100th digit is',a)
b=(N%100)//10   #b is the 10th place of the number
print('10th digit is',b)
c=N%10          #c is the 1st place of the number
print('1st digit is',c)
```

```python
#Write a code to find out the critical slope as a boolean statement
roll=57
min_slope=2
max_slope=6
slope=roll/10
if(slope>=min_slope)and(slope<=max_slope):
 print('The slope is critical and the value is',slope)
else:
 print('The slope is not critical and the value is', slope)
```

```python
#Find the area (cross section and surface) and volume of a sphere
r=157               #r is the radius of the sphere in cm
A1=3.1416*r**2      #A1 is the cross sectional area of the sphere in sq. cm
print(f'Cross Sectional Area is {A1:.2f} sq. cm')
A2=4*3.1416*r**2    #A2 is the surface area of the sphere in sq. cm
print(f'Surface Area is {A2:.2f} sq. cm')
V=(4/3)*3.1416*r**3 #V is the volume of the sphere in cubic cm
print(f'Volume is {V:.2f} cubic cm')
```
