```python
#Check whether a triangle is possible to form with the given sides or not, calculate the area if triangle is possible
a=2 #possible arm of a triangle in cm
b=3 #possible arm of a triangle in cm
c=3 #possible arm of a triangle in cm
if(a+b)>c and (b+c)>a and (c+a)>b:
  print('Triangle can be formed')
  s=(a+b+c)/2 #perimeter of the triangle in cm
  A=s*((s-a)*(s-b)*(s-c))**.5 #Area of the triangle in sq. cm
  print(f'The area is {A:.2f}')
else:
  print('Triangle cannot be formed')
```

```python
a=2
b=7
c=5
d=b**2-4*a*c
x=(-b+d**.5)/2*a
y=(-b-d**.5)/2*a
if(d>=0):
  print('The roots are real')
  print(f'The roots are {x:.2f} and {y:.2f}')
elif(d==0):
  print('There is only one root and it is',x)
else:
  print('The roots are not real')
```
